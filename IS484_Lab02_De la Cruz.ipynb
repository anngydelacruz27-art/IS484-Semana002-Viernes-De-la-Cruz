## Representación de un problema mediante estados

**Parte A: Completar la cha del problema para el siguiente caso: un repartidor debe ir desde el Almacén hasta la Zona Norte, pudiendo pasar por distintos puntos intermedios de la ciudad.**

|Estado| Descripción | 
|-----------|-----------|
|*Estado inicial:*   | El repartidor se encuentra en el almacen   |
| *Estado objetivo:*  | El repartidor ha llegado a la zona norte  | 
| *Acciones posibles:*   | El repartidor se puede desplazar en cualquier punto que este conectado a el  |
| *Costo de cada acción:*  | El costo de accion para los tramos va ser el mismo  |

**Parte B: Representar el mapa de la ciudad como un grafo en Python, usando un diccionario de listas (cada clave es un punto, cada valor es la lista de puntos conectados directamente):**

```python
mapa = {
    "Almacen": ["Centro", "Terminal"],
    "Centro": ["Almacen", "ZonaNorte", "Mercado"],
    "Terminal": ["Almacen", "Mercado"],
    "Mercado": ["Centro", "Terminal", "ZonaNorte"],
    "ZonaNorte": ["Centro", "Mercado"],
}
print(mapa)
```

**Implementación de BFS**

```python
def bfs(grafo, inicio, objetivo):
    """
    Busca el camino mas corto entre 'inicio' y 'objetivo' usando BFS.
    """
    cola = [[inicio]]   # Cola de caminos, empieza con el estado inicial
    visitados = set()   # Conjunto de nodos ya explorados, para no repetirlos y evitar ciclos
    while cola:
        # Para repetir mientras queden caminos por revisar en la cola
        camino = cola.pop(0)      # sacar el primer camino de la cola (FIFO)
        nodo_actual = camino[-1]  # El nodo actual es el ultimo del camino
        
        if nodo_actual == objetivo: # Si llegamos al objetivo, devuelve el camino completo y terminar.
            return camino 
            
        if nodo_actual not in visitados: # Expande el nodo si aun no fue explorado
            visitados.add(nodo_actual)   # marca como visitado 
            for vecino in grafo[nodo_actual]: # Recorre los vecinos directos del nodo actual
                nuevo_camino = camino + [vecino] #Crea un camino NUEVO para copiar del camino actual + el vecino al final
                cola.append(nuevo_camino)  # Agrega al FINAL de la cola. No se explora todavia y espera su turno detras de los caminos que ya estaban.

    return None # no se encontro un camino

resultado = bfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado:", resultado)
```

**Vectores y matrices con NumPy**

**Parte A: Representar cuatro observaciones de un dataset simple (por ejemplo, sensores de
temperatura y humedad) como un vector cada una, y juntarlas en una matriz:**

```python
import numpy as np
#Cada observacion es un VECTOR con 2 caracteristicas: [temperatura, humedad]
# np.array convierte la lista de Python en un arreglo de NumPy,
# que permite hacer operaciones matematicas (restar, multiplicar, etc.)

obs1 = np.array([22.5, 60]) # temperatura, humedad
obs2 = np.array([19.0, 45]) 
obs3 = np.array([25.3, 70])
obs4 = np.array([21.0, 50])

#Juntar los 4 vectores en una MATRIZ: cada fila es una observacion y cada columna es una caracteristica 
dataset = np.array([obs1, obs2, obs3, obs4])
print(dataset) #muestra la matriz
print("Forma de la matriz (filas, columnas):", dataset.shape) #shape devuelve (filas, columnas). Aqui: (4, 2) = 4 observaciones, 2 caracteristicas
```

**Producto escalar y distancia euclidiana**

**Parte A: Calcular el producto escalar entre dos observaciones:**

```python
#En este codigo multiplica las componentes que estan en la misma posicion
# y suma los resultados.
# (22.5 * 19.0) + (60 * 45) = 427.5 + 2700 = 3127.5
producto = np.dot(obs1, obs2) 
print("Producto escalar entre obs1 y obs2:", producto)
```

**Parte B: Calcular la distancia euclidiana entre dos observaciones, primero con la fórmula manual
y luego con NumPy:**

```python
# Formula manual
# obs1 - obs2:resta componente a componente: [3.5, 15.0]
# (...) ** 2:eleva cada diferencia al cuadrado: [12.25, 225.0]
# np.sum(...):suma los cuadrados: 237.25
# np.sqrt(...): raiz cuadrada: 15.40
distancia_manual = np.sqrt(np.sum((obs1 - obs2) ** 2))
print("Distancia manual:", distancia_manual)

# Con la funcion de NumPy
# np.linalg.norm calcula la longitud del vector diferencia,
# que es exactamente la misma formula ya implementada
distancia_numpy = np.linalg.norm(obs1 - obs2)
print("Distancia con NumPy:", distancia_numpy)
```

## Implementación de DFS

```python
def dfs(grafo, inicio, objetivo):
    #Define la funcion dfs. Recibe 3 datos:
    #grafo como diccionario, inicio como nodo de partida, objetivo como el nodo al que se quiere llegar
    
    pila = [[inicio]] # crea la pila y guarda caminos completos
    visitados = set() # conjunto vacio donde se anotan los nodos ya explorados y evitar caer a ciclos.
    
    while pila:
    #Repite mientras la pila tenga caminos por revisar.
        
        camino = pila.pop() # Esta es la diferencia con BFS, que usa pop(0) y saca el primero.
        # Por eso DFS sigue un camino hasta el fondo antes de probar otro.
        
        nodo_actual = camino[-1] #camino[-1] es el ultimo elemento de la lista: el nodo donde estamos parados.

        if nodo_actual == objetivo: #si ya se llego a la meta
            return camino #retorna el camino completo y termina la funcion

        if nodo_actual not in visitados: #Solo se expande el nodo si NO fue explorado antes.
            visitados.add(nodo_actual) #Lo marca como visitado para no volver a expandirlo.
            
            for vecino in grafo[nodo_actual]:
            # grafo[nodo_actual] busca en el diccionario la lista de vecinos directos.
            # El for los recorre de uno en uno.
                nuevo_camino = camino + [vecino] # Crea un camino NUEVO: copia del actual + el vecino al final
                pila.append(nuevo_camino) #Pone el camino nuevo al final de la pila.

    return None
resultado_dfs = dfs(mapa, "Almacen", "ZonaNorte")
print("Camino encontrado (DFS):", resultado_dfs)
```

**¿Son el mismo camino? ¿Cuál tiene menos pasos?**

- No son el mismo camino. BFS encontró `Almacen → Centro → ZonaNorte` (2 pasos) y DFS encontró `Almacen → Terminal → Mercado → ZonaNorte` (3 pasos).
- BFS tiene menos pasos, porque usa una cola (FIFO) y revisa primero todos los vecinos directos antes de alejarse.
- DFS usa una pila (LIFO): como `Terminal` entró después que `Centro`, sale primero y se sigue ese camino hasta el fondo. Llega a la meta, pero no por el camino más corto.
